```python
df['charges'] = round(df['charges'] , 2)
```

```python
x = df.drop('charges' , axis = 1)
y = df['charges']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test , y_train , y_test = train_test_split(x , y , test_size=0.2 , random_state= 40)
```

```python
x_train
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encode = OneHotEncoder(sparse_output= False , drop= 'first')
```

```python
x_train_enco = encode.fit_transform(x_train.select_dtypes(include= 'object'))
x_test_enco = encode.transform(x_test.select_dtypes(include='object'))
```

```python
x_train_dataset = pd.DataFrame(x_train_enco , columns= encode.get_feature_names_out())
x_test_dataset = pd.DataFrame(x_test_enco , columns= encode.get_feature_names_out())
```

```python
X_train = pd.concat([x_train_dataset , x_train.reset_index().select_dtypes(exclude= 'object')] , axis = 1)
X_test = pd.concat([x_test_dataset , x_test.reset_index().select_dtypes(exclude= 'object')] , axis = 1)
```

```python
X_test
```

```python
X_train.drop('index' , axis = 1 , inplace= True)
X_test.drop('index' , axis = 1 , inplace= True)
```

```python
X_train
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
y_train_scaled = scaler.fit_transform(y_train.values.reshape(-1,1))
y_test_scaled = scaler.transform(y_test.values.reshape(-1,1))
```

```python
X_train
```

```python
y_train
```

```python
y_train = pd.DataFrame(y_train_scaled , columns= ['charges'])
y_test = pd.DataFrame(y_test_scaled , columns= ['charges'])
```

```python
import torch
from torch import nn
```

```python
X_train_tensor = torch.tensor(X_train.values , dtype= torch.float32)
X_test_tensor = torch.tensor(X_test.values, dtype= torch.float32)
```

```python
y_train_tensor = torch.tensor(y_train.values , dtype= torch.float32)
y_test_tensor = torch.tensor(y_test.values ,dtype= torch.float32)
```

```python
class LinearRegressionModel(nn.Module):
  def __init__(self):
    super().__init__()
    self.linear = nn.Linear(in_features=8 , out_features=1)
  def forward(self , x):
    return self.linear(x)
```

```python
model = LinearRegressionModel()
loss_fn = torch.nn.MSELoss();
optimizer = torch.optim.SGD(model.parameters() , lr = 0.0001)
```

```python
for i in range(0, 10000):
  model.train()
  optimizer.zero_grad()
  outputs = model(X_train_tensor)
  loss = loss_fn(outputs , y_train_tensor)
  loss.backward()
  optimizer.step()
  if i % 10 == 0:
    print(f'Epoch: [{i+1}/10000] , Loss: {loss}')
```
