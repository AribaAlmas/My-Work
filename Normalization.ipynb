```python
df = pd.read_csv('wdbc.data', header = None)
df.head()
```

```python
df.head()
```

```python
df.replace(['?' , 'na' ,'nil' ' '] , np.nan , inplace=True)
```

```python
df
```

```python
df.isna().sum()
```

```python
df.dropna(axis=1)
```

```python
df_cleaned = df.dropna(axis=0)
df_cleaned
```

```python
df_cleaned
```

```python
df_nums = df_cleaned.select_dtypes(include='number')
```

```python
df_nums
```

```python
normalize(df_nums , norm = 'l2')
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
print(scaler.fit(df_nums))
```

```python
print(scaler.mean_)
```

```python
print(scaler.scale_)
```

```python
print(scaler.transform(df_nums))
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
minmax = MinMaxScaler()
```

```python
print(minmax.fit(df_nums))
```

```python
print(minmax.data_max_)
```

```python
print(minmax.transform(df_nums))
```
