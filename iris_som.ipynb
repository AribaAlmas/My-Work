```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('iris_data.csv')
```

```python
df.replace(['?',' ',' nil','na','-'],np.nan,inplace=True)
```

```python
x=df.iloc[:,:-1]
y=df.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,LabelEncoder
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=.25, random_state=33,stratify=y)
```

```python
df.columns
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scale = StandardScaler()
```

```python
x_train_norm = scale.fit_transform(x_train)
x_test_norm = scale.transform(x_test)
```

```python
print(x_train.shape)
```

```python
from minisom import MiniSom
```

```python
som = MiniSom(x=20 , y=20, input_len = 4, sigma = 1.0, learning_rate = 0.05 )
```

```python
som.random_weights_init(x_train_norm)
som.train_random(x_train_norm, num_iteration=1000)
```

```python
for x in x_test_norm:
    o = som.winner(x)
    print(f"BMU : {o}")
    
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(20,20))
for i,x in enumerate(x_test_norm):
    w = som.winner(x)
    plt.text(w[0]+0.5 , w[1]+0.5 , str(y_test[i]), color = plt.cm.tab10(y_test[i]), fontdict={'weight' : 'bold' , 'size' : 24})
plt.xlim([0,20])
plt.ylim([0,20])
plt.title("Minisom test data Pistachio label Map")
plt.grid()
plt.show()
 
```
